```python
def character_census(text):
    """
    Scan the text one character at a time and count the total number of
    characters, letters, digits, whitespace characters and 'other'
    characters (anything not covered by the previous three groups).
    """
    total = len(text)
    letters = 0
    digits = 0
    whitespace = 0
    other = 0

    for char in text:
        if char.isalpha():
            letters += 1
        elif char.isdigit():
            digits += 1
        elif char.isspace():
            whitespace += 1
        else:
            other += 1

    return {
        "total": total,
        "letters": letters,
        "digits": digits,
        "whitespace": whitespace,
        "other": other,
    }
```

```python
character_census('We are group 15')
```

## Task2

-------------------------
Case and vowel breakdown
--------------------------
Extend your program so that you also count:
1.	uppercase letters and lowercase letters separately,
2.	vowels and consonants separately (treat	as vowels, in either case),
3.	how many times each individual vowel appears. Add these to your report.


## Task2 Solution

```python
Block_of_text = input("Enter the text: ")

UC = 0
LC = 0
V = 0
CS = 0

a_count = 0
e_count = 0
i_count = 0
o_count = 0
u_count = 0

for ch in Block_of_text:
    if ch.isalpha():

        if ch.isupper():
            UC += 1
        elif ch.islower():
            LC += 1

        if ch.lower() in "aeiou":
            V += 1

            if ch.lower() == "a":
                a_count += 1
            elif ch.lower() == "e":
                e_count += 1
            elif ch.lower() == "i":
                i_count += 1
            elif ch.lower() == "o":
                o_count += 1
            elif ch.lower() == "u":
                u_count += 1
        else:
            CS += 1

print("Uppercase letters:", UC)
print("Lowercase letters:", LC)
print("Vowels:", V)
print("a:", a_count)
print("e:", e_count)
print("i:", i_count)
print("o:", o_count)
print("u:", u_count)
print("Consonants:", CS)
```

## Task3

----------------
Word statistics
----------------
Treat a word as a run of one or more letters (you may also allow an apostrophe inside a word, so that	and	count as single words). Anything that is not part of a word (spaces, punctuation, digits, newlines) separates one word from the next.
Using this rule, report:
1.	the total number of words,
2.	the longest word and its length (if several words tie for longest, report the first one),
3.	the average word length, rounded to one decimal place.


## Task3 Solution

```python

words = []
current_word = ""
length = len(text)
index = 0

while index < length:
    char = text[index]

    if char.isalpha():
        current_word += char
        index += 1
    elif (
        char == "'"
        and current_word != ""
        and index + 1 < length
        and text[index + 1].isalpha()
    ):
        # Apostrophe flanked by a letter on each side -> part of the word
        current_word += char
        index += 1
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""
        index += 1

# Catch a word that runs right up to the end of the text
if current_word != "":
    words.append(current_word)

total_words = len(words)

longest_word = ""
for word in words:
    if len(word) > len(longest_word):
        longest_word = word

if total_words > 0:
    total_letters_in_words = 0
    for word in words:
        total_letters_in_words += len(word)
    average_length = round(total_letters_in_words / total_words, 1)
else:
    average_length = 0.0

    return {
        "total_words": total_words,
        "longest_word": longest_word,
        "longest_word_length": len(longest_word),
        "average_length": average_length,
    }
```

## Task4

---------------------------
Line and sentence analysis
---------------------------
Add three more measurements:
1.	the number of lines in the text (a line is separated by the newline character '	'),
2.	the number of sentences, where a sentence ends with: , , or
3.	the length of the longest line (in characters).


## Task4 Solution

```python


# 1. Number of lines
lines = []
line = ""

for ch in text:
    if ch == "\n":
        lines.append(line)
        line = ""
    else:
        line += ch
lines.append(line)  # last line has no trailing \n

line_count = len(lines)

# 2. Number of sentences
sentence_count = 0
for ch in text:
    if ch == "." or ch == "!" or ch == "?":
        sentence_count += 1

# 3. Length of the longest line
longest_line = 0
for ln in lines:
    if len(ln) > longest_line:
        longest_line = len(ln)

print("--- Task 4: Lines & Sentences ---")
print("Number of lines    :", line_count)
print("Number of sentences:", sentence_count)
print("Longest line length:", longest_line)
```
