Line and sentence analysis
---------------------------
Add three more measurements:
1.	the number of lines in the text (a line is separated by the newline character '	'),
2.	the number of sentences, where a sentence ends with: , , or
3.	the length of the longest line (in characters).


## Task4 Solution

```python


# 1. Number of lines
lines = []
line = ""

for ch in text:
    if ch == "\n":
        lines.append(line)
        line = ""
    else:
        line += ch
lines.append(line)  # last line has no trailing \n

line_count = len(lines)

# 2. Number of sentences
sentence_count = 0
for ch in text:
    if ch == "." or ch == "!" or ch == "?":
        sentence_count += 1

# 3. Length of the longest line
longest_line = 0
for ln in lines:
    if len(ln) > longest_line:
        longest_line = len(ln)

print("--- Task 4: Lines & Sentences ---")
print("Number of lines    :", line_count)
print("Number of sentences:", sentence_count)
print("Longest line length:", longest_line)
```
